# FPL Historical Data Ingestion
This notebook focuses on fetching historical FPL data from https://github.com/vaastav/Fantasy-Premier-League/ and storing it in CSV format for efficient processing in subsequent notebooks.

## 1. Install Required Libraries
We need the `requests` library to interact with the FPL API, `pandas` for data manipulation, and `pyarrow` for Parquet file handling.

```python
import requests
import os
from datetime import datetime
```

## 2. Download FPL data
* The `download_fpl_gws_merged_data` function sends a GET request to download merged gameweek data from the github repo.
* The `download_fpl_team_data` function sends a GET request to download team data from the github repo.
* The `download_fpl_fixture_data` function sends a GET request to download fixture data from the github repo.
* The `download_fpl_player_data` function sends a GET request to download player data from the github repo.

```python
def download_fpl_merged_gw_data(start_year, end_year):
    """
    Downloads FPL data for each season from the specified URL.

    Args:
        start_year (int): The starting year of the season (e.g., 2016).
        end_year (int): The ending year of the season (e.g., 2025).
    """

    base_url = "https://raw.githubusercontent.com/vaastav/Fantasy-Premier-League/master/data/{season_year}/gws/merged_gw.csv"
    data_source = "fpl_historical"
    current_datetime = datetime.now().strftime("%Y%m%d_%H%M%S")
    data_dir = f"../data/raw/{data_source}/merged_gw/{current_datetime}/"

    # Create the data directory if it doesn't exist
    if not os.path.exists(data_dir):
        os.makedirs(data_dir)

    for year in range(start_year, end_year):
        season_year = f"{year}-{str(year + 1)[-2:]}"
        url = base_url.format(season_year=season_year)
        filename = f"{season_year.replace('-','_')}_merged_gw.csv"
        filepath = os.path.join(data_dir, filename)

        try:
            print(f"Downloading data for {season_year}...")
            response = requests.get(url)
            response.raise_for_status()  # Raise an exception for bad status codes

            with open(filepath, "w") as f:
                f.write(response.text)
            print(f"Data for {season_year} downloaded successfully to {filepath}")

        except requests.exceptions.RequestException as e:
            print(f"Error downloading data for {season_year}: {e}")

download_fpl_merged_gw_data(2022, 2024)
```
